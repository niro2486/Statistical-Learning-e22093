# Q. Analytical Derivation

###1.
Define the filter model
\begin{align*}
x^{-}_k &= A_{k-1}\,x^{+}_{k-1} + G_{k-1}\,w_{k-1}, \\
y^{-}_k &= H_k\,x^{-}_k + z_k,
\end{align*}
where
$$x^{+}_{k-1} \sim \mathscr{N}(m_{k-1}, P_{k-1}),$$


Because the filter equation is linear and the process noise is Gaussian, the predicted state $x^{-}_k$ is also Gaussian. Taking the expectation of the state equation yields the predicted mean:


$$m_k^- \triangleq \mathbb{E}[x^{-}_k] = A_{k-1}\mathbb{E}[x^{+}_{k-1}] + G_{k-1}\mathbb{E}[w_{k-1}] = A_{k-1}m_{k-1},$$


since $\mathbb{E}[w_{k-1}] = 0$.

The predicted covariance $P_k^-$ is computed by applying the variance operator to the state equation:
\begin{align*}
P_k^- &\triangleq \text{Var}(x^{-}_k) \\
&= A_{k-1}\text{Var}(x^{+}_{k-1})A_{k-1}^T + G_{k-1}\text{Var}(w_{k-1})G_{k-1}^T \\
&= A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T.
\end{align*}
Thus, prior to incorporating the new measurement, our belief of the state is characterized by the prior distribution:


$$x_k^- \sim \mathscr{N}(m_k^-, P_k^-).$$


###2.
Prior to its physical realization, the upcoming measurement at time step $k$ is treated as the random variable $y_k$. Based on our prior state belief, its expected value is:


$$\mathbb{E}[y^{-}_k] = H_k \mathbb{E}[x_k^-] + \mathbb{E}[z_k] = H_k m_k^-,$$


and its variance is:


$$\text{Var}(y^{-}_k) = H_k \text{Var}(x_k^-) H_k^T + \text{Var}(z_k) = H_k P_k^- H_k^T + \Sigma_m.$$

Therefore:
$$y_k^- \sim \mathscr{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m).$$

###3.

The cross-covariance between the predicted state $x_k^-$ and the anticipated measurement random variable $y_k$ is evaluated as:


$$\text{Cov}(x_k^-, y^{-}_k) = \text{Cov}(x_k^-, H_k x_k^- + z_k) = P_k^- H_k^T.$$

Therefore, the joint distribution of the predicted state and the upcoming measurement is a block-structured multivariate Gaussian:
$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathscr{N}\left(
\begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix},
\begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix}
\right).$$


###4.
At time step $k$, a physical measurement is observed, causing the random variable to take a specific numerical realization: $y_k = y^{\mathrm{obs}}_{k}$.

Applying the standard conditioning properties of multivariate normal distributions, we update our prior belief of the state by slicing the joint distribution at the realization $y^{\mathrm{obs}}_{k}$. This yields the posterior state distribution:


$$x^{+}_k\triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) \sim \mathscr{N}(m_k, P_k),$$


where the updated mean $m_k$ and updated covariance $P_k$ are given by:
\begin{align*}
K_k &\triangleq P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}, \\
m_k &= m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-), \\
P_k &= (I - K_k H_k) P_k^-.
\end{align*}


###5.

To find the conditional expectation $E[x^-_k \mid y^-_k = y^{obs}_k]$ and the conditional variance $Var(x^-_k \mid y^-_k = y^{obs}_k)$, we use the properties of conditional distributions for multivariate Gaussian variables.

From Question 3, we have the joint Gaussian distribution:
$$
\begin{bmatrix} x^-_k \\ y^-_k \end{bmatrix} \sim \mathcal{N} \left( \begin{bmatrix} m^-_k \\ H_k m^-_k \end{bmatrix}, \begin{bmatrix} P^-_k & P^-_k H^T_k \\ H_k P^-_k & H_k P^-_k H^T_k + \Sigma_m \end{bmatrix} \right)
$$

For any two jointly Gaussian random vectors $X$ and $Y$ partitioned as:
$$
\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathcal{N}\left( \begin{bmatrix} \mu_X \\ \mu_Y \end{bmatrix}, \begin{bmatrix} \Sigma_{XX} & \Sigma_{XY} \\ \Sigma_{YX} & \Sigma_{YY} \end{bmatrix} \right)
$$
The conditional distribution of $X$ given $Y = y$ is also Gaussian, with mean and variance given by the standard conditioning formulas:
1. **Conditional Mean:** $\mu_{X|Y} = \mu_X + \Sigma_{XY} \Sigma_{YY}^{-1} (y - \mu_Y)$
2. **Conditional Variance:** $\Sigma_{X|Y} = \Sigma_{XX} - \Sigma_{XY} \Sigma_{YY}^{-1} \Sigma_{YX}$

---

#### **1. Finding the Conditional Expectation (Mean)**
By substituting the corresponding blocks from our joint distribution into the conditional mean formula, we get:

$$
E[x^-_k \mid y^-_k = y^{obs}_k] = m^-_k + \left( P^-_k H^T_k \right) \left( H_k P^-_k H^T_k + \Sigma_m \right)^{-1} (y^{obs}_k - H_k m^-_k)
$$

Letting the Kalman Gain be $K_k \triangleq P^-_k H^T_k (H_k P^-_k H^T_k + \Sigma_m)^{-1}$, this simplifies to the standard measurement update equation for the mean:
$$
E[x^-_k \mid y^-_k = y^{obs}_k] = m^-_k + K_k (y^{obs}_k - H_k m^-_k)
$$

#### **2. Finding the Conditional Variance (Covariance)**
Similarly, substituting the blocks into the conditional variance formula gives:

$$
Var(x^-_k \mid y^-_k = y^{obs}_k) = P^-_k - \left( P^-_k H^T_k \right) \left( H_k P^-_k H^T_k + \Sigma_m \right)^{-1} \left( H_k P^-_k \right)
$$

Using the Kalman Gain $K_k$ defined above, we can rewrite this as the standard covariance update equation:
$$
Var(x^-_k \mid y^-_k = y^{obs}_k) = P^-_k - K_k H_k P^-_k = (I - K_k H_k)P^-_k
$$

# Q. 1-D Example

###1.


Given the state transition model $x_k = a x_{k-1} + w_{k-1}$ where the process noise is $w_{k-1} \sim \mathcal{N}(0, q)$, we want to find the predictive (prior) mean and variance of the state at step $k$, given all observations up to step $k-1$ (denoted as $Y_{k-1}$).

**1. Predictive Mean ($m_k^-$):**
The predictive mean is the expected value of $x_k$ given $Y_{k-1}$:
$$m_k^- = \mathbb{E}[x_k \mid Y_{k-1}]$$

Substitute the state equation:
$$m_k^- = \mathbb{E}[a x_{k-1} + w_{k-1} \mid Y_{k-1}]$$

By the linearity of expectation, and knowing that the process noise $w_{k-1}$ is zero-mean and independent of past observations:
$$m_k^- = a \mathbb{E}[x_{k-1} \mid Y_{k-1}] + \mathbb{E}[w_{k-1}]$$

Since the posterior mean from the previous step is defined as $m_{k-1} = \mathbb{E}[x_{k-1} \mid Y_{k-1}]$ and $\mathbb{E}[w_{k-1}] = 0$:
$$m_k^- = a m_{k-1}$$

**2. Predictive Variance ($P_k^-$):**
The predictive variance is the variance of $x_k$ given $Y_{k-1}$:
$$P_k^- = \text{Var}(x_k \mid Y_{k-1})$$

Substitute the state equation:
$$P_k^- = \text{Var}(a x_{k-1} + w_{k-1} \mid Y_{k-1})$$

Because the state $x_{k-1}$ and the process noise $w_{k-1}$ are independent, the variance of their sum is the sum of their variances. Also, pulling a constant $a$ out of a variance squares it:
$$P_k^- = a^2 \text{Var}(x_{k-1} \mid Y_{k-1}) + \text{Var}(w_{k-1})$$

Since the posterior variance from the previous step is $P_{k-1} = \text{Var}(x_{k-1} \mid Y_{k-1})$ and the process noise variance is $q$:
$$P_k^- = a^2 P_{k-1} + q$$

###2.
### **Background Setup**

From the prediction step (time update), we have the prior distribution of the state $x_k$ given all previous observations up to time $k-1$ ($Y_{k-1}$):
$$x_k | Y_{k-1} \sim \mathcal{N}(m_k^-, P_k^-)$$

The measurement equation is given as:
$$y_k = h x_k + z_k, \quad z_k \sim \mathcal{N}(0, r)$$

To find the posterior distribution (the measurement update), we need to compute the conditional distribution of $x_k$ given the new observation $y_k^{\text{obs}}$. Because the system is linear and Gaussian, the joint distribution of $x_k$ and $y_k$ given $Y_{k-1}$ is also Gaussian.

---

### **Step 1: Joint Distribution of State and Measurement**

We first find the joint Gaussian distribution:
$$\begin{pmatrix} x_k \\ y_k \end{pmatrix} \Bigg| Y_{k-1} \sim \mathcal{N} \left( \begin{pmatrix} \mu_x \\ \mu_y \end{pmatrix}, \begin{pmatrix} \Sigma_{xx} & \Sigma_{xy} \\ \Sigma_{yx} & \Sigma_{yy} \end{pmatrix} \right)$$

Let's calculate the means and covariances:

* **Mean of $x_k$:** $\mu_x = \mathbb{E}[x_k | Y_{k-1}] = m_k^-$
* **Mean of $y_k$:** $\mu_y = \mathbb{E}[h x_k + z_k | Y_{k-1}] = h\mathbb{E}[x_k | Y_{k-1}] + \mathbb{E}[z_k] = hm_k^-$
* **Variance of $x_k$:** $\Sigma_{xx} = \text{Var}(x_k | Y_{k-1}) = P_k^-$
* **Variance of $y_k$ (Innovation Covariance, $S_k$):**
    $$S_k = \Sigma_{yy} = \text{Var}(h x_k + z_k | Y_{k-1}) = h^2 \text{Var}(x_k | Y_{k-1}) + \text{Var}(z_k) = h^2 P_k^- + r$$
* **Cross-Covariance between $x_k$ and $y_k$:**
    $$\Sigma_{xy} = \text{Cov}(x_k, h x_k + z_k | Y_{k-1}) = h \text{Var}(x_k | Y_{k-1}) + \text{Cov}(x_k, z_k) = h P_k^-$$
    *(Note: $\Sigma_{yx} = \Sigma_{xy} = h P_k^-$)*

---

### **Step 2: Apply the Conditional Gaussian Formula**

For a joint Gaussian distribution, the conditional mean and variance of $x$ given $y$ are calculated using the standard formulas:
* **Conditional Mean:** $\mu_{x|y} = \mu_x + \Sigma_{xy}\Sigma_{yy}^{-1}(y - \mu_y)$
* **Conditional Variance:** $\Sigma_{x|y} = \Sigma_{xx} - \Sigma_{xy}\Sigma_{yy}^{-1}\Sigma_{yx}$

By substituting our specific variables into these formulas, we evaluate $x_k$ conditioned on the specific observation $y_k = y_k^{\text{obs}}$.

#### **Part 1: Deriving the Posterior Mean ($m_k$)**
Substitute the terms into the conditional mean formula:
$$m_k = \mu_{x_k | y_k^{\text{obs}}}$$
$$m_k = m_k^- + (h P_k^-)(S_k)^{-1}(y_k^{\text{obs}} - h m_k^-)$$

Let's define the following terms to match the notation in the question:
* **Innovation:** $v_k = y_k^{\text{obs}} - h m_k^-$
* **Kalman Gain:** $K_k = \frac{\Sigma_{xy}}{\Sigma_{yy}} = \frac{P_k^- h}{S_k}$

Substitute $v_k$ and $K_k$ back into the mean equation:
$$m_k = m_k^- + K_k v_k = m_k^- + \frac{P_k^- h}{S_k} (y_k^{\text{obs}} - h m_k^-)$$


#### **Part 2: Deriving the Posterior Variance ($P_k$)**
Substitute the terms into the conditional variance formula:
$$P_k = \Sigma_{x_k | y_k^{\text{obs}}}$$
$$P_k = P_k^- - (h P_k^-)(S_k)^{-1}(h P_k^-)$$
$$P_k = P_k^- - \frac{(P_k^-)^2 h^2}{S_k}$$

Factor out $P_k^-$ from the right-hand side:
$$P_k = \left( 1 - \frac{P_k^- h^2}{S_k} \right) P_k^-$$

Notice that the Kalman gain $K_k$ multiplied by $h$ gives us the exact fraction inside the parentheses:
$$K_k h = \left( \frac{P_k^- h}{S_k} \right) h = \frac{P_k^- h^2}{S_k}$$

Substituting $K_k h$ back into the equation yields:
$$P_k = (1 - K_k h) P_k^- = \left( 1 - \frac{P_k^- h^2}{S_k} \right) P_k^-$$


###3.
The predictive measurement distribution describes the expected distribution of $y_k$ *before* the actual observation is made, relying only on $Y_{k-1}$.

Using the measurement equation $y_k = h x_k + v_k$, we find the expected mean:
$$\mathbb{E}[y_k \mid Y_{k-1}] = \mathbb{E}[h x_k + v_k \mid Y_{k-1}] = h \mathbb{E}[x_k \mid Y_{k-1}] + \mathbb{E}[v_k]$$
$$\mathbb{E}[y_k \mid Y_{k-1}] = h m_k^- + 0 = h m_k^-$$

Next, we calculate the variance. Because the state $x_k$ and the measurement noise $v_k$ are independent:
$$\text{Var}(y_k \mid Y_{k-1}) = \text{Var}(h x_k + v_k \mid Y_{k-1})$$
$$\text{Var}(y_k \mid Y_{k-1}) = h^2 \text{Var}(x_k \mid Y_{k-1}) + \text{Var}(v_k)$$
$$\text{Var}(y_k \mid Y_{k-1}) = h^2 P_k^- + r$$

Since linear transformations of Gaussian variables remain Gaussian, the distribution is:
$$p(y_k \mid Y_{k-1}) = \mathcal{N}(h m_k^-, h^2 P_k^- + r)$$

###4.
The posterior-predictive measurement distribution describes the distribution of a replicated measurement $y_k$ *after* the state has been updated with the actual observation $y_k^{\text{obs}}$ (i.e., given $Y_k$).

Given $Y_k$, the updated state follows the posterior distribution $x_k \mid Y_k \sim \mathcal{N}(m_k, P_k)$. For a measurement drawn from this updated state, $y_k = h x_k + v_k$:

The posterior predictive mean is:
$$\mathbb{E}[y_k \mid Y_k] = \mathbb{E}[h x_k + v_k \mid Y_k] = h \mathbb{E}[x_k \mid Y_k] + \mathbb{E}[v_k]$$
$$\mathbb{E}[y_k \mid Y_k] = h m_k + 0 = h m_k$$

The posterior predictive variance is:
$$\text{Var}(y_k \mid Y_k) = \text{Var}(h x_k + v_k \mid Y_k) = h^2 \text{Var}(x_k \mid Y_k) + \text{Var}(v_k)$$
$$\text{Var}(y_k \mid Y_k) = h^2 P_k + r$$

Therefore, the distribution is:
$$p(y_k \mid Y_k) = \mathcal{N}(h m_k, h^2 P_k + r)$$

###5.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from scipy.stats import norm
from IPython.display import HTML, display

# ==========================================
# 1. Parameter Initialization
# ==========================================
a = 1.0    # State transition scalar
q = 0.5    # Process noise variance
h = 1.0    # Measurement scalar
r = 2.0    # Measurement noise variance

num_steps = 30 # Number of time steps to animate

# Initial prior state distribution N(m0, P0)
m_init = 0.0
P_init = 2.0

# ==========================================
# 2. Simulate True State and Measurements
# ==========================================
np.random.seed(42)
true_x = np.zeros(num_steps)
obs_y = np.zeros(num_steps)

# Generate initial true state
true_x[0] = np.random.normal(m_init, np.sqrt(P_init))
obs_y[0] = h * true_x[0] + np.random.normal(0, np.sqrt(r))

# Simulate the rest of the steps
for k in range(1, num_steps):
    true_x[k] = a * true_x[k-1] + np.random.normal(0, np.sqrt(q))
    obs_y[k] = h * true_x[k] + np.random.normal(0, np.sqrt(r))

# ==========================================
# 3. Setup the Plot and Animation
# ==========================================
fig, ax = plt.subplots(figsize=(10, 6))
x_axis = np.linspace(-10, 10, 500)

# Line objects for the distributions
line_prior, = ax.plot([], [], 'r--', lw=2, label='Prior Distribution (Predict)')
line_post, = ax.plot([], [], 'b-', lw=2, label='Posterior Distribution (Update)')
vline_true = ax.axvline(0, color='g', linestyle=':', lw=2, label='True State')
vline_obs = ax.axvline(0, color='orange', linestyle='-.', lw=2, label='Observation (y/h)')

ax.set_xlim(-10, 10)
ax.set_ylim(0, 1.0)
ax.set_title('Scalar Linear-Gaussian Filter: Prior vs Posterior')
ax.set_xlabel('State Space (x)')
ax.set_ylabel('Probability Density')
ax.legend(loc='upper right')
ax.grid(True, alpha=0.3)

# Variables to hold the filter state
m_current = m_init
P_current = P_init

def init():
    line_prior.set_data([], [])
    line_post.set_data([], [])
    return line_prior, line_post, vline_true, vline_obs

def update(frame):
    global m_current, P_current

    y_k = obs_y[frame]
    true_x_k = true_x[frame]

    # ---------------------------------------
    # Step 1: Predict (Calculate Prior)
    # ---------------------------------------
    m_prior = a * m_current
    P_prior = (a**2) * P_current + q

    # Generate PDF for the prior
    pdf_prior = norm.pdf(x_axis, m_prior, np.sqrt(P_prior))

    # ---------------------------------------
    # Step 2: Update (Calculate Posterior)
    # ---------------------------------------
    # Innovation (v_k) and Innovation Covariance (S_k)
    v_k = y_k - h * m_prior
    S_k = P_prior * (h**2) + r

    # Kalman Gain (K_k)
    K_k = P_prior * h / S_k

    # Posterior state
    m_current = m_prior + K_k * v_k
    P_current = (1 - K_k * h) * P_prior

    # Generate PDF for the posterior
    pdf_post = norm.pdf(x_axis, m_current, np.sqrt(P_current))

    # ---------------------------------------
    # Update Plot Lines
    # ---------------------------------------
    line_prior.set_data(x_axis, pdf_prior)
    line_post.set_data(x_axis, pdf_post)
    vline_true.set_xdata([true_x_k, true_x_k])

    # Scale observation to state space for visualization
    vline_obs.set_xdata([y_k / h, y_k / h])

    ax.set_title(f'Step {frame+1}/{num_steps} | Prior Var: {P_prior:.2f}, Post Var: {P_current:.2f}')

    return line_prior, line_post, vline_true, vline_obs

# Create the animation
ani = FuncAnimation(fig, update, frames=num_steps, init_func=init, blit=True, interval=800, repeat=False)

# Close the static figure so it doesn't display twice
plt.close(fig)

# Display the animation as an interactive HTML/JS object in Colab
display(HTML(ani.to_jshtml()))
```

## Part A

Assuming a sampling interval $\Delta t$, the constant-velocity kinematic equations are

$$
p_x(k) = p_x(k-1) + \Delta t\,v_x(k-1),
$$

$$
p_y(k) = p_y(k-1) + \Delta t\,v_y(k-1),
$$

$$
v_x(k) = v_x(k-1),
$$

$$
v_y(k) = v_y(k-1).
$$

Therefore, in matrix form,

$$
x_k =
\begin{bmatrix}
1 & 0 & \Delta t & 0\\
0 & 1 & 0 & \Delta t\\
0 & 0 & 1 & 0\\
0 & 0 & 0 & 1
\end{bmatrix}
x_{k-1}
+
G w_{k-1}.
$$

Thus,

$$
A =
\begin{bmatrix}
1 & 0 & \Delta t & 0\\
0 & 1 & 0 & \Delta t\\
0 & 0 & 1 & 0\\
0 & 0 & 0 & 1
\end{bmatrix}.
$$

---

Since the measurement contains only the position components, we have

$$
y_k =
\begin{bmatrix}
p_x(k)\\
p_y(k)
\end{bmatrix}
+
z_k.
$$

Equivalently,

$$
y_k = Hx_k + z_k,
$$

where

$$
H =
\begin{bmatrix}
1 & 0 & 0 & 0\\
0 & 1 & 0 & 0
\end{bmatrix}.
$$

The measurement noise is modeled as

$$
z_k \sim \mathscr{N}(0,\Sigma_m),
$$

with

$$
\Sigma_m = r^2 I_2.
$$

Here, $r$ is the standard deviation of the measurement noise in each position coordinate.

---


A common way to model uncertainty in a constant-velocity system is to assume that the unmodeled acceleration is random. Let

$$
w_{k-1} =
\begin{bmatrix}
a_x(k-1)\\
a_y(k-1)
\end{bmatrix},
$$

where

$$
w_{k-1} \sim \mathscr{N}(0,q^2I_2).
$$

Here, $q$ controls the acceleration-noise intensity.

Over one sampling interval $\Delta t$, the random acceleration affects both position and velocity:

$$
p_x(k)
=
p_x(k-1)
+
\Delta t\,v_x(k-1)
+
\frac{1}{2}\Delta t^2 a_x(k-1),
$$

$$
p_y(k)
=
p_y(k-1)
+
\Delta t\,v_y(k-1)
+
\frac{1}{2}\Delta t^2 a_y(k-1),
$$

$$
v_x(k)
=
v_x(k-1)
+
\Delta t\,a_x(k-1),
$$

$$
v_y(k)
=
v_y(k-1)
+
\Delta t\,a_y(k-1).
$$

Therefore,

$$
G =
\begin{bmatrix}
\frac{1}{2}\Delta t^2 & 0\\
0 & \frac{1}{2}\Delta t^2\\
\Delta t & 0\\
0 & \Delta t
\end{bmatrix},
$$

```python
"""
gps_kalman_filter.py

Filters a sequence of noisy 2D GPS position measurements using a
constant-velocity Kalman filter, with an optional Rauch-Tung-Striebel
(RTS) smoother for offline (whole-trajectory-available) use cases.

Why a Kalman filter:
- GPS noise is roughly zero-mean and uncorrelated between fixes.
- The true motion is smooth: position and velocity at time t are
  strongly related to position and velocity at time t-1.
A Kalman filter exploits both facts optimally (in the minimum-mean-
squared-error sense, under linear-Gaussian assumptions) by predicting
the next state from the motion model, then correcting that prediction
with the new noisy measurement, weighted by their relative uncertainty.

State vector:        x = [px, py, vx, vy]^T
Measurement vector:  z = [px_gps, py_gps]^T   (GPS reports position only)
"""

import numpy as np


class GPSKalmanFilter:
    """Constant-velocity Kalman filter for 2D GPS position data."""

    def __init__(self, dt, process_noise_std=0.5, gps_noise_std=5.0,
                 init_state=None, init_cov=None):
        """
        Parameters
        ----------
        dt : float
            Time between consecutive measurements (seconds).
        process_noise_std : float
            Standard deviation of the unmodeled acceleration (m/s^2).
            Larger = filter trusts the motion model less / reacts faster
            to real maneuvers, but smooths noise less.
        gps_noise_std : float
            Standard deviation of the GPS position noise (meters).
            Larger = filter trusts each individual fix less.
        init_state : array-like, shape (4,), optional
            Initial [px, py, vx, vy]. Defaults to zeros.
        init_cov : array-like, shape (4, 4), optional
            Initial state covariance. Defaults to a large uninformative value.
        """
        self.dt = dt

        # State transition matrix (constant velocity model):
        # px_k = px_{k-1} + vx_{k-1} * dt, etc.
        self.F = np.array([
            [1, 0, dt, 0],
            [0, 1, 0, dt],
            [0, 0, 1, 0],
            [0, 0, 0, 1],
        ])

        # Measurement matrix: GPS observes position only, not velocity
        self.H = np.array([
            [1, 0, 0, 0],
            [0, 1, 0, 0],
        ])

        # Process noise covariance: discrete white-noise-acceleration model.
        # Treats the unmodeled acceleration as white noise with std
        # process_noise_std, and derives how that propagates into
        # position/velocity uncertainty over one step of length dt.
        q = process_noise_std ** 2
        dt2, dt3, dt4 = dt**2, dt**3, dt**4
        self.Q = q * np.array([
            [dt4/4, 0,     dt3/2, 0],
            [0,     dt4/4, 0,     dt3/2],
            [dt3/2, 0,     dt2,   0],
            [0,     dt3/2, 0,     dt2],
        ])

        # Measurement noise covariance (assume independent x/y GPS noise)
        self.R = np.eye(2) * (gps_noise_std ** 2)

        # Initial state estimate and uncertainty
        self.x = np.zeros(4) if init_state is None else np.asarray(init_state, dtype=float)
        self.P = np.eye(4) * 100.0 if init_cov is None else np.asarray(init_cov, dtype=float)

    def predict(self):
        """Project the state forward by one time step using the motion model."""
        self.x = self.F @ self.x
        self.P = self.F @ self.P @ self.F.T + self.Q
        return self.x.copy(), self.P.copy()

    def update(self, z):
        """Correct the predicted state using a new noisy GPS measurement z = [px, py]."""
        z = np.asarray(z, dtype=float)
        y = z - self.H @ self.x                    # innovation (measurement residual)
        S = self.H @ self.P @ self.H.T + self.R     # innovation covariance
        K = self.P @ self.H.T @ np.linalg.inv(S)    # Kalman gain
        self.x = self.x + K @ y
        I = np.eye(self.P.shape[0])
        self.P = (I - K @ self.H) @ self.P
        return self.x.copy(), self.P.copy()

    def filter_sequence(self, measurements):
        """
        Run predict+update over an entire sequence of GPS fixes.

        Parameters
        ----------
        measurements : array-like, shape (N, 2)

        Returns
        -------
        states, covs : filtered state/covariance at each step, shape (N,4), (N,4,4)
        pred_states, pred_covs : the pre-update predictions at each step
            (kept around because the RTS smoother needs them)
        """
        measurements = np.asarray(measurements, dtype=float)
        n = len(measurements)

        states = np.zeros((n, 4))
        covs = np.zeros((n, 4, 4))
        pred_states = np.zeros((n, 4))
        pred_covs = np.zeros((n, 4, 4))

        for i, z in enumerate(measurements):
            x_pred, P_pred = self.predict()
            pred_states[i], pred_covs[i] = x_pred, P_pred
            x_upd, P_upd = self.update(z)
            states[i], covs[i] = x_upd, P_upd

        return states, covs, pred_states, pred_covs

    def rts_smooth(self, states, covs, pred_states, pred_covs):
        """
        Rauch-Tung-Striebel backward smoothing pass.

        Only valid offline (you need the entire sequence already filtered
        forward). Uses information from *later* points to improve *earlier*
        estimates, which a causal forward-only filter cannot do. Typically
        cuts RMSE substantially versus the forward filter alone.
        """
        n = len(states)
        smoothed = states.copy()
        smoothed_covs = covs.copy()

        for k in range(n - 2, -1, -1):
            P_pred_next = pred_covs[k + 1]
            C = covs[k] @ self.F.T @ np.linalg.inv(P_pred_next)
            smoothed[k] = states[k] + C @ (smoothed[k + 1] - pred_states[k + 1])
            smoothed_covs[k] = covs[k] + C @ (smoothed_covs[k + 1] - P_pred_next) @ C.T

        return smoothed, smoothed_covs


# ---------------------------------------------------------------------------
# Demo on synthetic data
# ---------------------------------------------------------------------------
if __name__ == "__main__":
    import matplotlib
    matplotlib.use("Agg")
    import matplotlib.pyplot as plt

    rng = np.random.default_rng(42)

    # --- Build a synthetic true trajectory (smooth curved path) ---
    dt = 1.0
    n_steps = 150
    t = np.arange(n_steps) * dt

    speed = 1.5
    true_px = speed * t * np.cos(0.04 * t)
    true_py = speed * t * np.sin(0.04 * t)

    # --- Simulate noisy GPS fixes ---
    gps_noise_std = 6.0  # meters, typical of consumer GPS
    noisy_px = true_px + rng.normal(0, gps_noise_std, n_steps)
    noisy_py = true_py + rng.normal(0, gps_noise_std, n_steps)
    measurements = np.column_stack([noisy_px, noisy_py])

    # --- Run the Kalman filter ---
    kf = GPSKalmanFilter(
        dt=dt,
        process_noise_std=0.3,
        gps_noise_std=gps_noise_std,
        init_state=[measurements[0, 0], measurements[0, 1], 0, 0],
        init_cov=np.diag([gps_noise_std**2, gps_noise_std**2, 10, 10]),
    )
    states, covs, pred_states, pred_covs = kf.filter_sequence(measurements)
    smoothed, smoothed_covs = kf.rts_smooth(states, covs, pred_states, pred_covs)

    # --- Evaluate ---
    def rmse(a, b):
        return np.sqrt(np.mean(np.sum((a - b) ** 2, axis=1)))

    true_pos = np.column_stack([true_px, true_py])
    print(f"RMSE (raw GPS):        {rmse(measurements, true_pos):.2f} m")
    print(f"RMSE (Kalman filter):  {rmse(states[:, :2], true_pos):.2f} m")
    print(f"RMSE (RTS smoothed):   {rmse(smoothed[:, :2], true_pos):.2f} m")

    # --- Plot ---
    fig, ax = plt.subplots(figsize=(8, 8))
    ax.plot(true_px, true_py, "k-", lw=2, label="True path")
    ax.scatter(noisy_px, noisy_py, s=15, c="red", alpha=0.5, label="Noisy GPS fixes")
    ax.plot(states[:, 0], states[:, 1], "b-", lw=2, label="Kalman filtered")
    ax.plot(smoothed[:, 0], smoothed[:, 1], "g--", lw=2, label="RTS smoothed")
    ax.set_xlabel("X position (m)")
    ax.set_ylabel("Y position (m)")
    ax.set_title("Filtering a noisy GPS trajectory")
    ax.legend()
    ax.set_aspect("equal")
    ax.grid(alpha=0.3)
    fig.tight_layout()
    fig.savefig("gps_filter_demo.png", dpi=150)
    print("Saved plot to gps_filter_demo.png")
```
